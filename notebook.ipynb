Todos os anos, estudantes do ensino médio nos Estados Unidos fazem o SAT, um exame padronizado que mede habilidades de leitura, matemática e escrita. São três seções — leitura, matemática e escrita — cada uma com **pontuação máxima de 800 pontos**. Essas provas são extremamente importantes para estudantes e universidades, pois têm um papel central no processo de admissão.

Analisar o desempenho das escolas é importante para diversos públicos, incluindo profissionais de políticas e educação, pesquisadores, governo e até pais que estão escolhendo em qual escola matricular seus filhos.

Você recebeu um conjunto de dados chamado `schools.csv`, com uma prévia abaixo.

Sua tarefa é responder a três perguntas essenciais sobre o desempenho no SAT das escolas públicas de Nova York (NYC).

```python
# Re-run this cell 
import pandas as pd

# Read in the data
schools = pd.read_csv("schools.csv")

# Preview the data
schools.head()

# Start coding here...
# Add as many cells as you like...
```

```python
best_math_schools = schools[schools['average_math']>= 640]
print (best_math_schools)
```

```python
best_math_schools = best_math_schools[['school_name', 'average_math']]
best_math_schools = best_math_schools.sort_values(by='average_math',ascending=False)

print(best_math_schools)
```

```python
schools['total_SAT'] = schools['average_math'] + schools['average_reading'] + schools['average_writing']
print (schools['total_SAT'])
```

```python
ord_total_SAT = schools.sort_values(by ='total_SAT', ascending = False)
ord_total_SAT.head(10)
```

```python
top_10_schools = ord_total_SAT[['school_name', 'total_SAT']].head(10)
print (top_10_schools)
```

```python
sch_per_borough = schools.groupby('borough')['total_SAT'].agg(['std', 'count', 'mean']).round(2)
print (sch_per_borough)
```

```python
sch_per_borough = sch_per_borough.sort_values(by='std', ascending=False)
print (sch_per_borough)
```

```python
large_std_dev = sch_per_borough.head(1)
print(large_std_dev)
```

```python
large_std_dev = large_std_dev.reset_index()
print(large_std_dev)
```

```python
largest_std_dev = large_std_dev.rename(columns= {'count':'num_schools', 'mean':'average_SAT', 'std':'std_SAT'}).round(2)
print (largest_std_dev)
```
